# Lesson 4: 用户主导型学习 Workflow

同一个 `thread_id` 会在当前 Jupyter 进程中持续保存 LangGraph 状态。

普通答疑不会写入文件，只有明确提出整理错题或总结复盘时才会执行对应操作。

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if project_root.name == "teacher":
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.facade import chat_v4

THREAD_ID = "lesson-4-demo"
```

```python
def run_v4_dialogue():
    results = []
    print("输入学习问题，或明确说‘整理错题’、‘总结复盘’。输入 /exit 结束。")
    while True:
        student_message = input("你：").strip()
        if student_message.lower() in {"/exit", "exit", "退出"}:
            return results
        if not student_message:
            print("请输入内容。")
            continue

        result = chat_v4(student_message, THREAD_ID)
        results.append(result)
        if result["error"]:
            print(f"错误：{result['error']}")
            continue

        print(f"教练：{result['text']}")
        print(f"等待：{result['waiting_for']}")
        if result["citations"]:
            print(f"知识引用：{result['citations']}")
        if result["tool_calls"]:
            print(f"工具调用：{result['tool_calls']}")
        if result["trace"]:
            print(f"最后步骤：{result['trace'][-1]}")
```

```python
v4_dialogue_results = run_v4_dialogue()
```
